```python
import os
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, Patch
from matplotlib.collections import LineCollection
from matplotlib.lines import Line2D

save_dir = "test_plots"
os.makedirs(save_dir, exist_ok=True)

# -------------------------------
# Dummy data
# -------------------------------
np.random.seed(0)
N = 600
threshold = 20.39

# Baseline PSNR around 21 (above threshold = "normal")
psnr_values = 21.0 + 0.25 * np.random.randn(N)

# GT anomaly segments (pink background regions)
gt_list = np.zeros(N, dtype=int)
gt_list[50:120]  = 1   # segment A
gt_list[200:260] = 1   # segment B
gt_list[400:470] = 1   # segment C
gt_list[520:540] = 1   # segment D (short)

# Inject PSNR drops to create the four cases:

# TRUE POSITIVE: PSNR drops inside GT segment A (pink + red line)
psnr_values[55:115] = 17.5 + 0.3 * np.random.randn(60)

# FALSE NEGATIVE: PSNR stays high inside GT segment B (pink + green line)
# -> leave as-is, it's still around 21

# TRUE POSITIVE partial: drop in second half of segment C
psnr_values[430:465] = 18.0 + 0.3 * np.random.randn(35)
# first half of C will be FN (pink + green)

# FALSE POSITIVE: PSNR drops outside any GT (white + red line)
psnr_values[300:340] = 19.0 + 0.3 * np.random.randn(40)

# A couple of brief dips crossing threshold for visual variety
psnr_values[480:490] = 19.8 + 0.2 * np.random.randn(10)

psnr_values = psnr_values.tolist()
gt_list = gt_list.tolist()

# -------------------------------
# Plot function (same as before)
# -------------------------------
def plot_video_result(video_name, psnr_values, gt_list, threshold):
    psnr_arr = np.array(psnr_values, dtype=float)
    gt_arr = np.array(gt_list)
    x = np.arange(len(psnr_arr))

    rect_start, rect_end = [], []
    active = False
    for i, val in enumerate(gt_arr):
        if val == 1 and not active:
            rect_start.append(i); active = True
        elif val == 0 and active:
            rect_end.append(i); active = False
    if active:
        rect_end.append(len(gt_arr))

    fig, ax = plt.subplots(figsize=(12, 5))

    y_min, y_max = psnr_arr.min(), psnr_arr.max()
    margin = (y_max - y_min) * 0.05
    for rs, re in zip(rect_start, rect_end):
        ax.add_patch(Rectangle(
            (rs, y_min - margin), re - rs, y_max - y_min + 2 * margin,
            facecolor='deeppink', alpha=0.30, edgecolor='none', zorder=1,
        ))

    points = np.array([x, psnr_arr]).T.reshape(-1, 1, 2)
    segments = np.concatenate([points[:-1], points[1:]], axis=1)
    below = psnr_arr[:-1] < threshold
    colors = np.where(below, '#d62728', '#2ca02c')
    lc = LineCollection(segments, colors=colors, linewidths=1.8, zorder=3)
    ax.add_collection(lc)

    ax.axhline(y=threshold, color='black', linestyle='--', linewidth=1.2, zorder=2)

    legend_handles = [
        Line2D([0], [0], color='#2ca02c', lw=2, label='PSNR (pred: normal)'),
        Line2D([0], [0], color='#d62728', lw=2, label='PSNR (pred: anomaly)'),
        Line2D([0], [0], color='black', lw=1.2, ls='--',
               label=f'Threshold ({threshold:.2f})'),
        Patch(facecolor='deeppink', alpha=0.30, label='GT anomaly'),
    ]

    ax.set_xlim(0, len(psnr_arr) - 1)
    ax.set_ylim(y_min - margin, y_max + margin)
    ax.set_xlabel('Frame')
    ax.set_ylabel('PSNR')
    ax.set_title(f'Video: {video_name} — GT (pink) vs Prediction (line color)')
    ax.legend(handles=legend_handles, loc='lower right', framealpha=0.9)
    plt.tight_layout()
    plt.savefig(os.path.join(save_dir, f"single_video_{video_name}.png"), dpi=150)
    plt.close()

plot_video_result("dummy", psnr_values, gt_list, threshold)
print(f"Saved: {save_dir}/single_video_dummy.png")
```
